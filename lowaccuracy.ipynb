```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split

# Define the mathematical functions
def func_Y1(x1, x2):
    numerator = 3 * x1**2 + 4 * x2 - 10
    denominator = 2 * x1**3 + x1 * x2
    return numerator / denominator

def func_Y2(x1, x2):
    return 3 * np.log(x1**2 + 2 * (x1 * x2)**2 + 5)

# Data generation function
def create_dataset(samples=12000):
    x1_values = np.random.uniform(-10, 20, samples)
    x2_values = np.random.uniform(-15, 30, samples)

    output_y1 = func_Y1(x1_values, x2_values)
    output_y2 = func_Y2(x1_values, x2_values)

    inputs = np.column_stack((x1_values, x2_values))
    outputs = np.column_stack((output_y1, output_y2))

    return inputs, outputs

# Generate and split the dataset
inputs, outputs = create_dataset()
x_train, x_test, y_train, y_test = train_test_split(inputs, outputs, test_size=0.2, random_state=0)

# Build neural network
neural_net = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(2,)),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(2)
])

# Compile model
neural_net.compile(optimizer='adam', loss='mean_squared_error')

# Train the neural network
training_history = neural_net.fit(x_train, y_train, epochs=80, batch_size=64, validation_split=0.15)

# Evaluate neural network
final_loss = neural_net.evaluate(x_test, y_test)
print(f'\nFinal Test Loss: {final_loss:.4f}')

# Predictions from neural network
predicted_outputs = neural_net.predict(x_test)

# Display sample predictions
for i in range(5):
    print(f'Input: {x_test[i]}, Prediction: {predicted_outputs[i]}, Actual: {y_test[i]}')
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Define the mathematical functions with safe operations
def func_Y1(x1, x2):
    denominator = 2 * x1**3 + x1 * x2
    denominator = np.where(np.abs(denominator) < 1e-5, 1e-5, denominator)
    numerator = 3 * x1**2 + 4 * x2 - 10
    return numerator / denominator

def func_Y2(x1, x2):
    return 3 * np.log(x1**2 + 2 * (x1 * x2)**2 + 5)

# Data generation function
def create_dataset(samples=20000):
    x1_values = np.random.uniform(-10, 20, samples)
    x2_values = np.random.uniform(-15, 30, samples)

    output_y1 = func_Y1(x1_values, x2_values)
    output_y2 = func_Y2(x1_values, x2_values)

    inputs = np.column_stack((x1_values, x2_values))
    outputs = np.column_stack((output_y1, output_y2))

    return inputs, outputs

# Generate and split the dataset
inputs, outputs = create_dataset()
x_train, x_test, y_train, y_test = train_test_split(inputs, outputs, test_size=0.2, random_state=0)

# Normalize data
input_scaler = StandardScaler()
output_scaler = StandardScaler()

x_train_scaled = input_scaler.fit_transform(x_train)
x_test_scaled = input_scaler.transform(x_test)
y_train_scaled = output_scaler.fit_transform(y_train)
y_test_scaled = output_scaler.transform(y_test)

# Build a more powerful neural network
neural_net = keras.Sequential([
    keras.layers.Dense(256, activation='relu', input_shape=(2,)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(2)
])

# Compile model with adjusted learning rate
neural_net.compile(optimizer=keras.optimizers.Adam(learning_rate=0.0005), loss='mean_squared_error')

# Train the neural network
training_history = neural_net.fit(
    x_train_scaled, y_train_scaled, epochs=200, batch_size=32, validation_split=0.2, verbose=1
)

# Evaluate neural network
final_loss = neural_net.evaluate(x_test_scaled, y_test_scaled)
print(f'\nFinal Test Loss: {final_loss:.4f}')

# Predictions from neural network (scaled back to original scale)
predicted_outputs_scaled = neural_net.predict(x_test_scaled)
predicted_outputs = output_scaler.inverse_transform(predicted_outputs_scaled)

# Display sample predictions
for i in range(5):
    print(f'Input: {x_test[i]}, Prediction: {predicted_outputs[i]}, Actual: {y_test[i]}')
```
