```python
import cv2
import mediapipe as mp
import numpy as np
from google.colab.patches import cv2_imshow
from IPython.display import display, Javascript
from base64 import b64decode

# Initialize MediaPipe
mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

class FingerCounter:
    def __init__(self):
        self.hands = mp_hands.Hands(
            static_image_mode=False,
            max_num_hands=2,
            min_detection_confidence=0.7,
            min_tracking_confidence=0.7
        )

        # نقاط landmark برای انگشتان
        self.tip_ids = [4, 8, 12, 16, 20]  # شست, اشاره, وسط, حلقه, کوچک
        self.pip_ids = [3, 6, 10, 14, 18]   # مفاصل انگشتان

    def count_fingers(self, hand_landmarks):
        """شمارش دقیق انگشتان"""
        fingers = []

        # شست - بررسی بر اساس موقعیت افقی
        if hand_landmarks.landmark[self.tip_ids[0]].x > hand_landmarks.landmark[self.tip_ids[0] - 1].x:
            fingers.append(1)  # شست باز
        else:
            fingers.append(0)  # شست بسته

        # چهار انگشت دیگر - بررسی بر اساس موقعیت عمودی
        for i in range(1, 5):
            if hand_landmarks.landmark[self.tip_ids[i]].y < hand_landmarks.landmark[self.pip_ids[i]].y:
                fingers.append(1)  # انگشت باز
            else:
                fingers.append(0)  # انگشت بسته

        return sum(fingers), fingers

    def get_finger_positions(self, hand_landmarks, frame_shape):
        """گرفتن موقعیت واقعی انگشتان روی تصویر"""
        positions = []
        height, width = frame_shape

        for tip_id in self.tip_ids:
            lm = hand_landmarks.landmark[tip_id]
            x = int(lm.x * width)
            y = int(lm.y * height)
            positions.append((x, y))

        return positions

    def detect_hands(self, frame):
        """تشخیص دست در فریم"""
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = self.hands.process(rgb_frame)
        return results

    def draw_detection(self, frame, results):
        """رسم تشخیص و اطلاعات"""
        if results.multi_hand_landmarks:
            for hand_landmarks in results.multi_hand_landmarks:
                # رسم landmarks
                mp_drawing.draw_landmarks(
                    frame,
                    hand_landmarks,
                    mp_hands.HAND_CONNECTIONS,
                    mp_drawing_styles.get_default_hand_landmarks_style(),
                    mp_drawing_styles.get_default_hand_connections_style()
                )

                # شمارش انگشتان
                total_fingers, finger_states = self.count_fingers(hand_landmarks)

                # گرفتن موقعیت انگشتان
                finger_positions = self.get_finger_positions(hand_landmarks, frame.shape)

                # نمایش تعداد انگشتان
                cv2.putText(frame, f"Fingers: {total_fingers}", (10, 50),
                           cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 255, 0), 3)

                # نمایش وضعیت هر انگشت
                finger_names = ["Thumb", "Index", "Middle", "Ring", "Pinky"]
                for i, (state, pos, name) in enumerate(zip(finger_states, finger_positions, finger_names)):
                    color = (0, 255, 0) if state == 1 else (0, 0, 255)
                    status = "Open" if state == 1 else "Closed"

                    # نمایش دایره روی نوک انگشتان
                    cv2.circle(frame, pos, 10, color, -1)

                    # نمایش وضعیت هر انگشت
                    cv2.putText(frame, f"{name}: {status}", (10, 100 + i * 30),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)

        return frame

def take_photo():
    """گرفتن عکس از وبکم در Colab"""
    js = Javascript('''
        async function takePhoto() {
            const div = document.createElement('div');
            const capture = document.createElement('button');
            capture.textContent = 'Capture';
            div.appendChild(capture);

            const video = document.createElement('video');
            video.style.display = 'block';
            const stream = await navigator.mediaDevices.getUserMedia({video: true});

            document.body.appendChild(div);
            div.appendChild(video);
            video.srcObject = stream;
            await video.play();

            // Resize the output to fit the video element.
            google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

            // Wait for Capture to be clicked.
            await new Promise((resolve) => capture.onclick = resolve);

            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);
            stream.getVideoTracks()[0].stop();
            div.remove();
            return canvas.toDataURL('image/jpeg', 0.8);
        }
    ''')

    display(js)
    data = eval_js('takePhoto()')
    binary = b64decode(data.split(',')[1])
    return binary

def process_image_from_webcam():
    """پردازش تصویر از وبکم در Colab"""
    print("📸 در حال راه اندازی وبکم...")
    print("روی دکمه Capture کلیک کنید تا عکس بگیرید")

    # گرفتن عکس از وبکم
    photo_bytes = take_photo()

    # تبدیل بایت‌ها به تصویر
    with open('webcam_photo.jpg', 'wb') as f:
        f.write(photo_bytes)

    # خواندن تصویر
    frame = cv2.imread('webcam_photo.jpg')

    if frame is None:
        print("❌ خطا در گرفتن عکس")
        return

    print("✅ عکس گرفته شد. در حال پردازش...")

    # ایجاد detector
    counter = FingerCounter()

    # تشخیص دست
    results = counter.detect_hands(frame)

    if results.multi_hand_landmarks:
        # رسم تشخیص و اطلاعات
        frame = counter.draw_detection(frame, results)
        print("✅ دست تشخیص داده شد!")
    else:
        # نمایش پیام عدم تشخیص
        cv2.putText(frame, "No Hand Detected", (10, 50),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
        cv2.putText(frame, "Try again with better lighting", (10, 100),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)
        print("❌ دست تشخیص داده نشد")

    # نمایش تصویر
    print("📊 نمایش نتیجه:")
    cv2_imshow(frame)

    # ذخیره نتیجه
    cv2.imwrite('result.jpg', frame)
    print("💾 نتیجه در فایل result.jpg ذخیره شد")

def process_uploaded_image():
    """پردازش تصویر آپلود شده"""
    from google.colab import files
    import io

    print("📁 لطفا یک عکس آپلود کنید:")
    uploaded = files.upload()

    if not uploaded:
        print("❌ هیچ فایلی آپلود نشد")
        return

    # خواندن اولین فایل آپلود شده
    file_name = list(uploaded.keys())[0]
    file_bytes = uploaded[file_name]

    # تبدیل بایت‌ها به تصویر
    frame = cv2.imdecode(np.frombuffer(file_bytes, np.uint8), cv2.IMREAD_COLOR)

    if frame is None:
        print("❌ خطا در خواندن تصویر")
        return

    print(f"✅ تصویر {file_name} آپلود شد. در حال پردازش...")

    # ایجاد detector
    counter = FingerCounter()

    # تشخیص دست
    results = counter.detect_hands(frame)

    if results.multi_hand_landmarks:
        # رسم تشخیص و اطلاعات
        frame = counter.draw_detection(frame, results)
        print("✅ دست تشخیص داده شد!")
    else:
        # نمایش پیام عدم تشخیص
        cv2.putText(frame, "No Hand Detected", (10, 50),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
        cv2.putText(frame, "Try another image", (10, 100),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)
        print("❌ دست تشخیص داده نشد")

    # نمایش تصویر
    print("📊 نمایش نتیجه:")
    cv2_imshow(frame)

    # ذخیره نتیجه
    cv2.imwrite('processed_result.jpg', frame)
    print("💾 نتیجه در فایل processed_result.jpg ذخیره شد")

# اجرای اصلی
if __name__ == "__main__":
    print("🤖 سیستم شمارش انگشتان در Google Colab")
    print("1. گرفتن عکس از وبکم")
    print("2. آپلود عکس از کامپیوتر")

    try:
        choice = input("لطفا انتخاب کنید (1 یا 2): ")
        if choice == "1":
            process_image_from_webcam()
        else:
            process_uploaded_image()
    except Exception as e:
        print(f"❌ خطا: {e}")
        print("از گزینه 2 (آپلود عکس) استفاده کنید")
        process_uploaded_image()
```

```python
import cv2
import mediapipe as mp
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files
from IPython.display import display, Javascript
from base64 import b64decode

# Initialize MediaPipe
mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

class FingerCounter:
    def __init__(self):
        self.hands = mp_hands.Hands(
            static_image_mode=True,  # تغییر به True برای عکس‌های ثابت
            max_num_hands=2,
            min_detection_confidence=0.5,
            min_tracking_confidence=0.5
        )

        # نقاط landmark برای انگشتان
        self.tip_ids = [4, 8, 12, 16, 20]  # شست, اشاره, وسط, حلقه, کوچک

    def count_fingers(self, hand_landmarks):
        """شمارش دقیق انگشتان"""
        fingers = []

        # شست - بررسی بر اساس موقعیت افقی
        if hand_landmarks.landmark[self.tip_ids[0]].x > hand_landmarks.landmark[3].x:
            fingers.append(1)  # شست باز
        else:
            fingers.append(0)  # شست بسته

        # چهار انگشت دیگر - بررسی بر اساس موقعیت عمودی
        for i in range(1, 5):
            if hand_landmarks.landmark[self.tip_ids[i]].y < hand_landmarks.landmark[self.tip_ids[i] - 2].y:
                fingers.append(1)  # انگشت باز
            else:
                fingers.append(0)  # انگشت بسته

        return sum(fingers), fingers

    def get_finger_positions(self, hand_landmarks, frame_shape):
        """گرفتن موقعیت واقعی انگشتان روی تصویر"""
        positions = []
        # تصحیح unpacking شکل تصویر
        if len(frame_shape) == 3:
            height, width, _ = frame_shape
        else:
            height, width = frame_shape[0], frame_shape[1]

        for tip_id in self.tip_ids:
            lm = hand_landmarks.landmark[tip_id]
            x = int(lm.x * width)
            y = int(lm.y * height)
            positions.append((x, y))

        return positions

    def detect_hands(self, frame):
        """تشخیص دست در فریم"""
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = self.hands.process(rgb_frame)
        return results

    def draw_detection(self, frame, results):
        """رسم تشخیص و اطلاعات"""
        if results.multi_hand_landmarks:
            for hand_landmarks in results.multi_hand_landmarks:
                # رسم landmarks
                mp_drawing.draw_landmarks(
                    frame,
                    hand_landmarks,
                    mp_hands.HAND_CONNECTIONS,
                    mp_drawing_styles.get_default_hand_landmarks_style(),
                    mp_drawing_styles.get_default_hand_connections_style()
                )

                # شمارش انگشتان
                total_fingers, finger_states = self.count_fingers(hand_landmarks)

                # گرفتن موقعیت انگشتان
                finger_positions = self.get_finger_positions(hand_landmarks, frame.shape)

                # نمایش تعداد انگشتان
                cv2.putText(frame, f"Fingers: {total_fingers}", (10, 50),
                           cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 255, 0), 3)

                # نمایش وضعیت هر انگشت
                finger_names = ["Thumb", "Index", "Middle", "Ring", "Pinky"]
                for i, (state, pos, name) in enumerate(zip(finger_states, finger_positions, finger_names)):
                    color = (0, 255, 0) if state == 1 else (0, 0, 255)
                    status = "Open" if state == 1 else "Closed"

                    # نمایش دایره روی نوک انگشتان
                    cv2.circle(frame, pos, 10, color, -1)

                    # نمایش وضعیت هر انگشت
                    cv2.putText(frame, f"{name}: {status}", (10, 100 + i * 30),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)

        return frame

def take_photo():
    """گرفتن عکس از وبکم در Colab"""
    js = Javascript('''
        async function takePhoto() {
            const div = document.createElement('div');
            const capture = document.createElement('button');
            capture.textContent = 'Capture';
            div.appendChild(capture);

            const video = document.createElement('video');
            video.style.display = 'block';
            const stream = await navigator.mediaDevices.getUserMedia({video: true});

            document.body.appendChild(div);
            div.appendChild(video);
            video.srcObject = stream;
            await video.play();

            google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

            await new Promise((resolve) => capture.onclick = resolve);

            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            canvas.getContext('2d').drawImage(video, 0, 0);
            stream.getVideoTracks()[0].stop();
            div.remove();
            return canvas.toDataURL('image/jpeg', 0.8);
        }
        takePhoto()
    ''')

    display(js)

    # منتظر ماندن برای نتیجه
    from IPython.display import clear_output
    import time
    time.sleep(5)  # زمان برای کلیک کردن

    # این تابع در Colab نیاز به تنظیمات بیشتری دارد، بهتر است از آپلود استفاده کنیم

def process_uploaded_image():
    """پردازش تصویر آپلود شده - بهترین گزینه برای Colab"""
    print("📁 لطفا یک عکس از دست خود آپلود کنید:")
    print("• دست را باز نگه دارید")
    print("• از نور کافی استفاده کنید")
    print("• پس‌زمینه ساده باشد")

    uploaded = files.upload()

    if not uploaded:
        print("❌ هیچ فایلی آپلود نشد")
        return

    # خواندن اولین فایل آپلود شده
    file_name = list(uploaded.keys())[0]
    file_bytes = uploaded[file_name]

    # تبدیل بایت‌ها به تصویر
    frame = cv2.imdecode(np.frombuffer(file_bytes, np.uint8), cv2.IMREAD_COLOR)

    if frame is None:
        print("❌ خطا در خواندن تصویر")
        return

    print(f"✅ تصویر '{file_name}' آپلود شد")
    print("🔍 در حال پردازش تصویر...")

    # ایجاد detector
    counter = FingerCounter()

    # تشخیص دست
    results = counter.detect_hands(frame)

    if results.multi_hand_landmarks:
        # رسم تشخیص و اطلاعات
        frame = counter.draw_detection(frame, results)
        print("✅ دست تشخیص داده شد!")
        print(f"📊 تعداد انگشتان تشخیص داده شده: {sum(counter.count_fingers(results.multi_hand_landmarks[0])[1])}")
    else:
        # نمایش پیام عدم تشخیص
        cv2.putText(frame, "No Hand Detected", (10, 50),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
        cv2.putText(frame, "Try another image with better lighting", (10, 100),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)
        print("❌ دست تشخیص داده نشد")
        print("💡 نکات: از نور بهتر استفاده کنید، دست را کاملا باز کنید، پس‌زمینه ساده باشد")

    # نمایش تصویر
    print("📊 نتیجه پردازش:")
    cv2_imshow(frame)

    # ذخیره نتیجه
    cv2.imwrite('finger_count_result.jpg', frame)
    print("💾 نتیجه در فایل 'finger_count_result.jpg' ذخیره شد")

def simple_demo():
    """دموی ساده با عکس نمونه"""
    print("🎯 در حال اجرای دمو با عکس نمونه...")

    # دانلود یک عکس نمونه از اینترنت
    import urllib.request
    import os

    # URL عکس نمونه دست
    sample_url = "https://upload.wikimedia.org/wikipedia/commons/thumb/5/51/Hand_Outline.svg/1200px-Hand_Outline.svg.png"

    try:
        # دانلود عکس
        urllib.request.urlretrieve(sample_url, "sample_hand.png")

        # خواندن عکس
        frame = cv2.imread("sample_hand.png")

        if frame is None:
            print("❌ خطا در خواندن عکس نمونه")
            return

        # تغییر اندازه اگر太大了
        frame = cv2.resize(frame, (500, 500))

        # ایجاد detector
        counter = FingerCounter()

        # تشخیص دست
        results = counter.detect_hands(frame)

        if results.multi_hand_landmarks:
            frame = counter.draw_detection(frame, results)
            print("✅ دست در عکس نمونه تشخیص داده شد!")
        else:
            cv2.putText(frame, "Sample Hand - No detection", (10, 50),
                       cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
            print("⚠️  در عکس نمونه دست تشخیص داده نشد (ممکن است طرح باشد)")

        # نمایش نتیجه
        cv2_imshow(frame)

    except Exception as e:
        print(f"❌ خطا در دمو: {e}")
        print("لطفا عکس دست خود را آپلود کنید")

# اجرای اصلی
print("🤖 سیستم شمارش انگشتان در Google Colab")
print("=" * 50)
print("1. آپلود عکس دست خود (توصیه می‌شود)")
print("2. مشاهده دمو با عکس نمونه")

try:
    choice = input("لطفا انتخاب کنید (1 یا 2): ").strip()
    if choice == "2":
        simple_demo()
    else:
        process_uploaded_image()
except Exception as e:
    print(f"❌ خطا: {e}")
    print("در حال اجرای گزینه پیشفرض (آپلود عکس)...")
    process_uploaded_image()
```
